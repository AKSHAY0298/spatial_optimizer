```python
import gurobipy as gp

# Create a new model
m = gp.Model("test")

# Add variables
x = m.addVar(vtype=gp.GRB.BINARY, name="x")
y = m.addVar(vtype=gp.GRB.BINARY, name="y")

# Set objective
m.setObjective(x + y, gp.GRB.MAXIMIZE)

# Optimize model
m.optimize()

print(f"Test optimization successful! Objective value: {m.ObjVal}")
```
